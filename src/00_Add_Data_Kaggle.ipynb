```python
import pandas as pd

# 1. Đọc file CSV
df = pd.read_csv("../datasets/raw/Symptom2Disease.csv")

# 2. Lọc các dòng có label = 'diabetes'
filtered_df = df[df['label'] == 'diabetes'][['label', 'text']]

# 3. Xuất ra file csv
output_file = "../datasets/raw/diabetes_filtered.csv"
with open(output_file, "w", encoding="utf-8") as f:
    for _, row in filtered_df.iterrows():
        f.write(f"{row['label']}\t{row['text']}\n")

print("Done! File saved as:", output_file)
```
