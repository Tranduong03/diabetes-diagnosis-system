```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Đọc dữ liệu
df = pd.read_csv(r"D:\IT\HK1_Y4\Class\Project_1\Project_2\datasets\raw\diabetes_structured.csv")

# Hiển thị tỷ lệ nhãn ban đầu
print("📊 Phân bố nhãn trước khi tách:")
print(df["Outcome"].value_counts())
print(df["Outcome"].value_counts(normalize=True))
```

```python
# Tách dữ liệu train/test (80% - 20%)
train_df, test_df = train_test_split(
  df, 
  test_size=0.2, 
  random_state=42, 
  shuffle=True,
  stratify=df['Outcome']  # Giữ tỷ lệ phân phối của nhãn 'Outcome'
)

# Hiển thị tỷ lệ nhãn sau khi tách
print("\n📊 Phân bố nhãn trong tập train:")
print(train_df["Outcome"].value_counts())
print(train_df["Outcome"].value_counts(normalize=True))

print("\n📊 Phân bố nhãn trong tập test:")
print(test_df["Outcome"].value_counts())
print(test_df["Outcome"].value_counts(normalize=True))
```

```python
# Đường dẫn lưu file
train_path = r"D:\IT\HK1_Y4\Class\Project_1\Project_2\datasets\processed\train_data.csv"
test_path = r"D:\IT\HK1_Y4\Class\Project_1\Project_2\datasets\processed\test_data.csv"

# Lưu ra file CSV
train_df.to_csv(train_path, index=False)
test_df.to_csv(test_path, index=False)
```

```python
print("✅ Đã tách dữ liệu thành công!")
print(f"Kích thước tập train: {train_df.shape}")
print(f"Kích thước tập test: {test_df.shape}")
```
