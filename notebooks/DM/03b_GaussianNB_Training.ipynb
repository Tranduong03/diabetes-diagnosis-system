## Lý do chọn thuật toán Gaussian Naive Bayes

**Gaussian Naive Bayes (GNB)** được lựa chọn vì các lý do sau:

- **Phù hợp với dữ liệu dạng liên tục:**  
  Bộ dữ liệu *Pima Indians Diabetes Database* có các thuộc tính như `Glucose`, `BloodPressure`, `BMI`,… là các giá trị **số thực (continuous)**.  
  Thuật toán GNB giả định rằng dữ liệu trong mỗi lớp tuân theo **phân phối chuẩn (Gaussian)**, nên rất phù hợp cho loại dữ liệu này.

- **Đơn giản, dễ huấn luyện và giải thích:**  
  GaussianNB dựa trên **định lý Bayes** và giả định **độc lập giữa các đặc trưng**, giúp việc tính toán xác suất nhanh, hiệu quả và dễ hiểu — đặc biệt quan trọng trong bài toán y học cần mô hình minh bạch.

- **Hiệu quả với dữ liệu nhỏ và vừa:**  
  Bộ dữ liệu có **1268 mẫu** và **8 đặc trưng**, do đó GNB huấn luyện nhanh, không cần tinh chỉnh nhiều siêu tham số.

- **Phù hợp bài toán phân loại nhị phân:**  
  Biến mục tiêu `Outcome` gồm:
  - `0`: Không bị tiểu đường  
  - `1`: Bị tiểu đường  
  ⇒ Đây là **bài toán phân loại nhị phân (binary classification)** — rất thích hợp cho GNB.

````python
# ===========================
# 03b_GaussianNB_Training.ipynb
# ===========================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, precision_score, recall_score, f1_score

# ===========================
# ĐỌC DỮ LIỆU
# ===========================
df1 = pd.read_csv("../../datasets/processed/diabetes_dataset1_mean.csv")
df2 = pd.read_csv("../../datasets/processed/diabetes_dataset2_median.csv")


def evaluate_model(y_true, y_pred, model_name, dataset_name):
    print(f"\n{'='*60}")
    print(f"{model_name} - {dataset_name}")
    print(f"{'='*60}")

    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred)
    rec = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    print(f"- Accuracy:  {acc:.4f}")
    print(f"- Precision: {prec:.4f}")
    print(f"- Recall:    {rec:.4f}")
    print(f"- F1-Score:  {f1:.4f}")
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, target_names=['Không bị', 'Bị']))

    # return acc, prec, rec, f1

# ===========================
# TRAIN & TEST GaussianNB
# ===========================
print("\n--- Gaussian NB với Dataset 1 (Mean) ---")
X1 = df1.drop('Outcome', axis=1)
y1 = df1['Outcome']
X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
nb1 = GaussianNB()
nb1.fit(X1_train, y1_train)
y1_probs = nb1.predict_proba(X1_test)[:, 1]
threshold_1 = 0.3
y1_pred_custom = (y1_probs >= threshold_1).astype(int)
evaluate_model(y1_test, y1_pred_custom, f"Gaussian NB (Threshold {threshold_1})", "Dataset 1 (Mean)")

print("\n--- Gaussian NB với Dataset 2 (Median) ---")
X2 = df2.drop('Outcome', axis=1)
y2 = df2['Outcome']
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=42)
nb2 = GaussianNB()
nb2.fit(X2_train, y2_train)
y2_probs = nb2.predict_proba(X2_test)[:, 1]
threshold_2 = 0.3
y2_pred_custom = (y2_probs >= threshold_2).astype(int)
evaluate_model(y2_test, y2_pred_custom, f"Gaussian NB (Threshold {threshold_2})", "Dataset 2 (Median)")
````
