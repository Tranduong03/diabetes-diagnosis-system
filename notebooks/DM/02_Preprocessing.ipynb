📖 [02] TIỀN XỬ LÝ DỮ LIỆU - PIMA INDIANS DIABETES

Mục tiêu: Làm sạch và chuẩn bị dữ liệu cho bước huấn luyện mô hình bằng cách xử lý giá trị thiếu, thay thế giá trị không hợp lệ và chuẩn hóa dữ liệu.

Các bước chính:

1. Đọc và kiểm tra dữ liệu

- Tải file diabetes_structured.csv.

- Kiểm tra kích thước, giá trị thiếu và số lượng giá trị 0 trong các cột quan trọng.

2. Làm sạch dữ liệu

- Thay thế các giá trị 0 không hợp lệ (ở các cột Glucose, BloodPressure, SkinThickness, Insulin, BMI) bằng giá trị NaN.

- Xác nhận lại số lượng giá trị thiếu sau khi thay thế.

3. Tạo các bộ dữ liệu xử lý khác nhau

- Dataset 1 – Mean: Thay thế các giá trị thiếu bằng mean của từng cột.

- Dataset 2 – Median: Thay thế các giá trị thiếu bằng median của từng cột.

- Dataset 3 – Scaled: Chuẩn hóa dữ liệu từ Dataset 2 bằng StandardScaler để đưa các giá trị về cùng thang đo.

4. Lưu dữ liệu đã xử lý

- Lưu ba file kết quả:

  - diabetes_dataset1_mean.csv

  - diabetes_dataset2_median.csv

  - diabetes_dataset3_scaled.csv

5. Kết luận

- Dữ liệu sau khi tiền xử lý không còn giá trị thiếu hoặc không hợp lệ.

- Bộ dữ liệu đã được chuẩn bị ở ba mức độ khác nhau để thử nghiệm với các mô hình học máy: Mean, Median và Scaled.

1.1. Import thư viện và đọc dữ liệu

```python
# ===========================
# ĐỌC DỮ LIỆU
# ===========================
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Đọc file dữ liệu
df = pd.read_csv('diabetes_structured.csv')

print(f"Kích thước dữ liệu: {df.shape[0]} dòng, {df.shape[1]} cột")
print(df.head())
print(df.info())
```

1.2. Kiểm tra dữ liệu ban đầu

```python
# ===========================
#  KIỂM TRA DỮ LIỆU
# ===========================
print("\nKiểm tra giá trị thiếu:")
print(df.isnull().sum())

print("\nKiểm tra số lượng giá trị 0 trong các cột:")
zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
for col in zero_cols:
    if col in df.columns:
        zero_count = (df[col] == 0).sum()
        print(f"  • {col:<20}: {zero_count:>3} giá trị 0 ({zero_count/len(df)*100:>5.2f}%)")
```

**Nhận xét:**

Các cột không có giá trị NaN, nhưng có nhiều giá trị bằng 0 ở các cột không hợp lý:

- Glucose: 5 giá trị 0

- BloodPressure: 35 giá trị 0

- SkinThickness: 227 giá trị 0

- Insulin: 374 giá trị 0

- BMI: 11 giá trị 0

Đây là dấu hiệu dữ liệu thiếu cần xử lý.

2. Làm sạch dữ liệu (thay 0 bằng NaN)

```python
# ===========================
# LÀM SẠCH DỮ LIỆU
# ===========================
df_clean = df.copy()

# Thay thế giá trị 0 bằng NaN trong các cột không thể có giá trị 0
for col in zero_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].replace(0, np.nan)

print("\nĐã thay thế giá trị 0 bằng NaN trong các cột:", zero_cols)
print("\nSố lượng giá trị thiếu sau khi xử lý:")
print(df_clean.isnull().sum())
```

**Nhận xét:**

Sau bước này, các giá trị 0 được thay bằng NaN:

- Glucose có 5 giá trị thiếu

- BloodPressure có 35 giá trị thiếu

- SkinThickness có 227 giá trị thiếu

- Insulin có 374 giá trị thiếu

- BMI có 11 giá trị thiếu

3.a. Xử lý giá trị thiếu bằng MEAN

```python
# ===========================
#  DATASET 1 - THAY GIÁ TRỊ THIẾU BẰNG MEAN
# ===========================
print("\n" + "-"*60)
print("DATASET 1: Thay giá trị thiếu bằng MEAN")
print("-"*60)

df_mean = df_clean.copy()
for col in zero_cols:
    if col in df_mean.columns:
        mean_value = df_mean[col].mean()
        df_mean[col] = df_mean[col].fillna(mean_value)
        print(f"  - {col}: Thay NaN bằng mean = {mean_value:.2f}")

print(f"Số lượng giá trị thiếu còn lại: {df_mean.isnull().sum().sum()}")

```

Tất cả giá trị thiếu đã được thay thế bằng giá trị
trung bình

Không còn giá trị thiếu.

3.b. Xử lý giá trị thiếu bằng MEDIAN

```python
# ===========================
# DATASET 2 - THAY GIÁ TRỊ THIẾU BẰNG MEDIAN
# ===========================
print("\n" + "-"*60)
print("DATASET 2: Thay giá trị thiếu bằng MEDIAN")
print("-"*60)

df_median = df_clean.copy()
for col in zero_cols:
    if col in df_median.columns:
        median_value = df_median[col].median()
        df_median[col] = df_median[col].fillna(median_value)
        print(f"  - {col}: Thay NaN bằng median = {median_value:.2f}")

print(f"Số lượng giá trị thiếu còn lại: {df_median.isnull().sum().sum()}")
```

Toàn bộ NaN được thay bằng trung vị.

Không còn giá trị thiếu. Phương pháp này giúp dữ liệu ổn định hơn khi có outlier.

3.c. Chuẩn hóa dữ liệu (StandardScaler)

```python
# ===========================
# DATASET 3 - CHUẨN HÓA DỮ LIỆU (TỪ DATASET 2)
# ===========================
print("\n" + "-"*60)
print("DATASET 3: Chuẩn hóa dữ liệu (từ Dataset 2)")
print("-"*60)

df_scaled = df_median.copy()

if 'Outcome' in df_scaled.columns:
    X_to_scale = df_scaled.drop('Outcome', axis=1)
    y = df_scaled['Outcome']
else:
    X_to_scale = df_scaled

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_to_scale)
X_scaled_df = pd.DataFrame(X_scaled, columns=X_to_scale.columns)

if 'Outcome' in df_scaled.columns:
    df_scaled_final = pd.concat([X_scaled_df, y.reset_index(drop=True)], axis=1)
else:
    df_scaled_final = X_scaled_df

print("Đã chuẩn hóa dữ liệu bằng StandardScaler!")
print(df_scaled_final.head())
```

**Nhận xét:**

Các giá trị đã được chuẩn hóa về trung bình ≈ 0, độ lệch chuẩn ≈ 1.

Dữ liệu này sẵn sàng để huấn luyện các mô hình như KNN

4. Lưu dữ liệu đã xử lý

```python
# ===========================
# LƯU DỮ LIỆU ĐÃ XỬ LÝ
# ===========================
print("\n" + "="*60)
print("LƯU DỮ LIỆU ĐÃ XỬ LÝ")
print("="*60)

df_mean.to_csv('diabetes_dataset1_mean.csv', index=False)
print("Đã lưu: diabetes_dataset1_mean.csv")

df_median.to_csv('diabetes_dataset2_median.csv', index=False)
print("Đã lưu: diabetes_dataset2_median.csv")

df_scaled_final.to_csv('diabetes_dataset3_scaled.csv', index=False)
print("Đã lưu: diabetes_dataset3_scaled.csv")

print("\nHOÀN THÀNH TIỀN XỬ LÝ DỮ LIỆU!")
```

**Nhận xét:**

Ba tệp dữ liệu mới được tạo ra trong thư mục hiện tại:

- diabetes_dataset1_mean.csv

- diabetes_dataset2_median.csv

- diabetes_dataset3_scaled.csv

Mỗi tệp tương ứng với một phương pháp tiền xử lý khác nhau, có thể sử dụng để huấn luyện và so sánh độ chính xác giữa các mô hình.

5. Kết luận

  Quá trình tiền xử lý dữ liệu bao gồm ba bước chính: làm sạch giá trị sai lệch, xử lý giá trị thiếu bằng hai phương pháp (trung bình và trung vị), và chuẩn hóa dữ liệu bằng StandardScaler.

  Các bước này giúp loại bỏ nhiễu, giảm sai số, đồng thời đảm bảo rằng dữ liệu đầu vào có chất lượng tốt, phù hợp cho các mô hình học máy.

  Kết quả thu được là ba phiên bản dữ liệu khác nhau, giúp so sánh và lựa chọn cách tiền xử lý tối ưu nhất cho bài toán dự đoán tiểu đường.
